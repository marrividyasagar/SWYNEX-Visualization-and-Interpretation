```python
import pandas as pd
import numpy as np
import glob
import os

# Locate uploaded dataset
files = glob.glob("/content/GSE20966*")

print("Files found:", files)

# Load dataset
file_path = files[0]

df = pd.read_csv(
    file_path,
    sep="\t",
    comment="!",
    index_col=0,
    compression="infer"
)

print("\nDataset loaded successfully!")

# Display dataset dimensions
print("Dataset shape:", df.shape)

# Display first five rows
display(df.head())
```

```python
# Check missing values
missing_values = df.isnull().sum()

print("Total missing values:", missing_values.sum())

print("\nMissing values in each sample:")
print(missing_values)

print("\nRows containing missing values:",
      df.isnull().any(axis=1).sum())
```

```python
print("Total rows:", len(df))

print("Duplicate probe identifiers:",
      df.index.duplicated().sum())

print("Unique probe identifiers:",
      df.index.nunique())
```

```python
print("Dataset information:")
df.info()

print("\nData types:")
print(df.dtypes.value_counts())
```

```python
summary = df.describe()

display(summary)
```

```python
# Calculate statistical measurements

mean_values = df.mean()
median_values = df.median()
std_values = df.std()

statistics = pd.DataFrame({
    "Mean": mean_values,
    "Median": median_values,
    "Standard Deviation": std_values
})

display(statistics.round(3))
```

```python
control_samples = df.columns[:10]
diabetes_samples = df.columns[10:]

control_mean = df[control_samples].mean(axis=1)
diabetes_mean = df[diabetes_samples].mean(axis=1)

comparison = pd.DataFrame({
    "Control_Mean": control_mean,
    "Diabetes_Mean": diabetes_mean
})

comparison["Difference"] = (
    comparison["Diabetes_Mean"] -
    comparison["Control_Mean"]
)

display(comparison.head(10))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 6))

sns.boxplot(data=df)

plt.title("Gene Expression Distribution Across 20 Samples")
plt.xlabel("Sample ID")
plt.ylabel("Expression Value")
plt.xticks(rotation=90)

plt.tight_layout()
plt.savefig("expression_boxplot.png", dpi=300)
plt.show()
```

```python
correlation = df.corr()

plt.figure(figsize=(12, 9))

sns.heatmap(
    correlation,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Between Gene Expression Samples")
plt.tight_layout()
plt.savefig("correlation_heatmap.png", dpi=300)
plt.show()
```

```python
import os

os.makedirs("results", exist_ok=True)

df.to_csv("results/cleaned_gene_expression.csv")
statistics.to_csv("results/sample_statistics.csv")
comparison.to_csv("results/control_vs_diabetes.csv")

print("All result files saved successfully!")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Transpose so samples become rows
X = df.T

# Standardize each probe across samples
X_scaled = StandardScaler().fit_transform(X)

# Perform PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(X_scaled)

# Create PCA dataframe
pca_df = pd.DataFrame(
    pca_result,
    columns=["PC1", "PC2"]
)

pca_df["Group"] = (
    ["Control"] * 10 + ["Diabetes"] * 10
)

# Plot
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=pca_df,
    x="PC1",
    y="PC2",
    hue="Group",
    s=100
)

plt.title("PCA of Gene Expression Samples")
plt.xlabel(
    f"PC1 ({pca.explained_variance_ratio_[0]*100:.2f}% variance)"
)
plt.ylabel(
    f"PC2 ({pca.explained_variance_ratio_[1]*100:.2f}% variance)"
)

plt.tight_layout()
plt.savefig("PCA_plot.png", dpi=300)
plt.show()
```

```python
import pandas as pd
import numpy as np

file_path = list(uploaded.keys())[0]

df = pd.read_csv(
    file_path,
    sep="\t",
    comment="!",
    index_col=0
)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Transpose so samples become rows
X = df.T

# Standardize each probe across samples
X_scaled = StandardScaler().fit_transform(X)

# Perform PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(X_scaled)

# Create PCA dataframe
pca_df = pd.DataFrame(
    pca_result,
    columns=["PC1", "PC2"]
)

pca_df["Group"] = (
    ["Control"] * 10 + ["Diabetes"] * 10
)

# Plot
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=pca_df,
    x="PC1",
    y="PC2",
    hue="Group",
    s=100
)

plt.title("PCA of Gene Expression Samples")
plt.xlabel(
    f"PC1 ({pca.explained_variance_ratio_[0]*100:.2f}% variance)"
)
plt.ylabel(
    f"PC2 ({pca.explained_variance_ratio_[1]*100:.2f}% variance)"
)

plt.tight_layout()
plt.savefig("PCA_plot.png", dpi=300)
plt.show()
```

```python
plt.figure(figsize=(12, 8))

sns.scatterplot(
    data=pca_df,
    x="PC1",
    y="PC2",
    hue="Group",
    s=100
)

for i, sample in enumerate(df.columns):
    plt.annotate(
        sample,
        (pca_df.loc[i, "PC1"], pca_df.loc[i, "PC2"]),
        fontsize=8,
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.title("PCA of Gene Expression Samples with Sample IDs")
plt.tight_layout()
plt.savefig("PCA_plot_labeled.png", dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Select 10 probes with highest variance
top_probes = df.var(axis=1).nlargest(10).index

# Calculate group averages
plot_data = pd.DataFrame({
    "Control": df.loc[top_probes, df.columns[:10]].mean(axis=1),
    "Diabetes": df.loc[top_probes, df.columns[10:]].mean(axis=1)
})

# Convert to long format
plot_data = plot_data.reset_index().melt(
    id_vars="ID_REF",
    var_name="Group",
    value_name="Mean Expression"
)

# Plot
plt.figure(figsize=(14, 7))

sns.barplot(
    data=plot_data,
    x="ID_REF",
    y="Mean Expression",
    hue="Group"
)

plt.title("Top 10 Variable Probes: Control vs Diabetes")
plt.xlabel("Probe ID")
plt.ylabel("Mean Expression Signal")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("group_comparison.png", dpi=300)
plt.show()
```

```python
# Examine the top 10 variable probes
probe_check = df.loc[top_probes]

probe_summary = pd.DataFrame({
    "Control_Mean": probe_check.iloc[:, :10].mean(axis=1),
    "Diabetes_Mean": probe_check.iloc[:, 10:].mean(axis=1),
    "Minimum": probe_check.min(axis=1),
    "Maximum": probe_check.max(axis=1),
    "Variance": probe_check.var(axis=1)
})

probe_summary["Difference"] = (
    probe_summary["Diabetes_Mean"] -
    probe_summary["Control_Mean"]
)

display(probe_summary.round(3))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Exclude the unusual probe only from this visualization
df_visual = df.drop(index="g13544062_3p_at")

# Select the 10 most variable remaining probes
top_probes_clean = df_visual.var(axis=1).nlargest(10).index

# Calculate group means
plot_data = pd.DataFrame({
    "Control": df_visual.loc[top_probes_clean].iloc[:, :10].mean(axis=1),
    "Diabetes": df_visual.loc[top_probes_clean].iloc[:, 10:].mean(axis=1)
})

# Prepare chart data
plot_data.index.name = "Probe ID"
plot_data = plot_data.reset_index().melt(
    id_vars="Probe ID",
    var_name="Group",
    value_name="Mean Expression"
)

# Plot
plt.figure(figsize=(14, 7))

sns.barplot(
    data=plot_data,
    x="Probe ID",
    y="Mean Expression",
    hue="Group"
)

plt.title("Control vs Diabetes: Top 10 Variable Probes")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("group_comparison_clean.png", dpi=300)
plt.show()
```
